```python
def CheckPassword(n):
    d = 0
    u = 0

    if len(n) < 4:
        return 0

    if n[0].isdigit():
        return 0

    for i in n:
        if i.isdigit():
            d = 1
        if i.isupper():
            u = 1
        if i == ' ' or i == '/':
            return 0

    if d == 1 and u == 1:
        return 1
    else:
        return 0


n = input()
print(CheckPassword(n))
```

```python
def FindAutoCount(n):
    if n is None:
        print(0)
        return

    for i in range(len(n)):
        if n.count(str(i)) != int(n[i]):
            print(0)
            return

    print(len(set(n)))


n = input()
FindAutoCount(n)
```

```python
n = input()

d = 0
s = ""
l = ""
w = 0

k = int(n)

while k > 0:
    d = k % 2
    s = str(d)+s
    k = k // 2


for i in s:
    if i == "0":
        l = l + "1"
    else:
        l = l + "0"

p = len(l) - 1

for i in range(len(l)):
    w = w + int(l[i]) * 2**p
    p = p - 1

print(w)
```
