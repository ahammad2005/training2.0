```python
def CheckPassword(n):
    d = 0
    u = 0

    if len(n) < 4:
        return 0

    if n[0].isdigit():
        return 0

    for i in n:
        if i.isdigit():
            d = 1
        if i.isupper():
            u = 1
        if i == ' ' or i == '/':
            return 0

    if d == 1 and u == 1:
        return 1
    else:
        return 0


n = input()
print(CheckPassword(n))
```

```python
def FindAutoCount(n):
    if n is None:
        print(0)
        return

    for i in range(len(n)):
        if n.count(str(i)) != int(n[i]):
            print(0)
            return

    print(len(set(n)))


n = input()
FindAutoCount(n)
```

```python
def strongPassword(s):
    l=False
    u=False
    d=False
    r=0
    for i in range(len(s)):
        if s[i].islower():
            l=True
        if s[i].isupper():
            u=True
        if s[i].isdigit():
            d=True
    for i in range(1, len(s) - 1):
        if s[i]==s[i-1] and s[i]==s[i+1]:
            r+=1
    m=0
    if not l:
        m+=1
    if not u:
        m+=1
    if not d:
        m+=1
    x=0
    if len(s)<6:
        x=6-len(s)

    ans=max(m, x, r)
    return ans

s=input()
print(strongPassword(s))
```
